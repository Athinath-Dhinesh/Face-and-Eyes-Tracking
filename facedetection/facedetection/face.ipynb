```python
import cv2

face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

print("Cascade loaded:", not face_cascade.empty())

```

```python
grey=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
cv2.imshow('grey',grey)
cv2.waitKey()   
cv2.destroyAllWindows()
```

```python
import matplotlib.pyplot as plt

plt.imshow(grey, cmap='grey')
plt.show()
```

```python
face=face_cascade.detectMultiScale(grey,1.1,4)
face
```

```python
conv_img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(conv_img)
```

```python
gray=cv2.cvtColor(img,cv2.COLOR_RGB2GRAY)
faces=face_cascade.detectMultiScale(gray,minNeighbors=4)
faces
```

```python
for (x,y,w,h) in faces:
    cv2.rectangle(conv_img,(x,y),(x+w,y+h),(255,0,0),2)
plt.imshow(conv_img)


```

```python
Video=cv2.VideoCapture('video.mp4')
while True:
    success,img=Video.read()
    gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
    faces=face_cascade.detectMultiScale(gray,minNeighbors=4)
    print(len(faces))
    cv2.imshow('video',img)
    if cv2.waitKey(1) & 0xFF==ord('q'):
        break
Video.release()
cv2.destroyAllWindows()
```
